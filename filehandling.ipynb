```python
with open('intro.txt') as f:
    print(f.read().split())
```

```python
with open('intro.txt') as f:
    words=f.read().split(',')
    print("Total words",len(words))
```

```python
with open('intro.txt','r')as f:
    words=f.read().split()
    words1=set(words)
    print('set',words)
    print("Total words",len(words1))
```

```python
with open('intro.txt','r') as f:
    word1=f.read().split()
    word_count={}
    for word in word1:
        if word in word_count:
            word_count[word]+=1
        else:
            word_count[word]=1
        

print(word_count)
```

```python
data={
    "name":"Aditya",
    "age":23,
    "course":"IT" 
    
}
```

```python
import json
with open('json1.json','w') as f:
    json.dump(data,f)
```

```python
import json
with open('json1.json','r+') as f:
    json.dump(data,f)
    f.seek(0)
    da = json.load(f)
    print(da)
    print(len(da))
```

```python
import csv
with open("data.csv","w",newline="") as file:
    writer=csv.writer(file)
    writer.writerow(["Name","Age","Course"])
    writer.writerow(["Aditya",22,"CSE"])
```

```python
row=[
    ["Name","Age"],
    ["Aman",21],
    ["Ravi",23]
]
with open("data.csv","w",newline="") as file:
    writer=csv.writer(file)
    writer.writerow(row)
```

```python
import csv
with open("data.csv","w",newline="")as file:
    fieldnames=["name","age"]
    writer=csv.DictWriter(file,fieldnames=fieldnames)
    writer.writeheader()
    writer.writerow({"name":"Aditya","age":45})
    writer.writerow({"name":"Aman","age":56})

```

```python
with open("data.csv","r") as file:
    reader =csv.DictReader(file)
    print(reader)#->Object address
    for row in reader:
        print(row)
```
