```python
data1={
    'Name':['A','B','C'],
    'age':[23,45,67],
    'salary':[10000,20000,30000]
}

data2={
    'Name':['A','BB','C'],
    'age':[23,54,67],
    'salary':[10000,25000,30000]
}
```

```python
new_df1=pd.DataFrame(data1)
new_df2=pd.DataFrame(data2)
```

```python
new_df1.head()
```

```python
new_df2.head()
```

```python
pd.concat([new_df1,new_df2],ignore_index=True)
```

```python
pd.concat([new_df1,new_df2],ignore_index=True,axis=1)
```

```python
new_df=pd.concat([new_df1,new_df2],ignore_index=True,axis=0)
new_df.head(6)
```

```python
new_df.drop_duplicates(inplace=True)
new_df.head()
```

#### **Merge**

```python
new_df1.head()
```

```python
new_df2.head()
```

```python
pd.merge(new_df1,new_df2)
```

```python
pd.merge(new_df1,new_df2,on='age')
```

```python
pd.merge(new_df1,new_df2,on='age',how='outer')
#outer join
```

```python
pd.merge(new_df1,new_df2,on='age',how='inner')
#inner join
```
