```python
x=np.arange(1,15)
y=np.sin(x)
z=np.cos(x)
print(x,y,z,sep='\n\n')
```

```python
plt.figure(figsize=(8,4))
plt.plot(x,y,label='SINQ',color='red')
plt.plot(x,z,label='COSQ')
plt.title("SINQ  & COSQ")
plt.xlabel('Data Points')
plt.ylabel('Association values')
plt.xticks()
plt.yticks()
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.title("SINQ  & COSQ")
plt.plot(x,y,label='SINQ',color='red')
plt.plot(x,z,label='COSQ')
plt.xlabel('Data Points')
plt.ylabel('Association values')
plt.legend()

plt .subplot(1,2,2)
plt.title("SINQ & COSQ2")
plt.plot(x,y,label='SINQ',color='green')
plt.plot(x,z,label='COSQ',color='red')
plt.xlabel('Data Points2')
plt.ylabel('Association values2')


plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,8))

plt.subplot(3,3,1)
plt.title("SINQ  & COSQ")
plt.plot(x,y,label='SINQ',color='red')
plt.plot(x,z,label='COSQ')
plt.xlabel('Data Points')
plt.ylabel('Association values')
plt.grid()
plt.legend(loc='upper right')

plt .subplot(3,3,5)
plt.title("SINQ & COSQ2")
plt.plot(x,y,label='SINQ',color='green')
plt.plot(x,z,label='COSQ',color='red')
plt.xlabel('Data Points2')
plt.ylabel('Association values2')
plt.grid()
plt.legend(loc='upper right')

plt .subplot(3,3,9)
plt.title("SINQ & COSQ2")
plt.plot(x,y,label='SINQ',color='blue')
plt.plot(x,z,label='COSQ',color='black')
plt.xlabel('Data Points3')
plt.ylabel('Association values3')
plt.subplots_adjust(wspace=0.8)
plt.grid()


plt.legend(loc='upper right')
plt.show()
```
