```python
s='code@'
lastch=s[-1:]
if lastch.isdigit():
    print("Ends with digit")
elif lastch.isalpha():
    print("Ends with character")
else:
    print("Ends with special character")
```

```python
text = 'Python'
if text[:1].isupper():
    print("start with uppercase")
elif text[-1:].islower():
    print("Ends with lowercase")
else:
    print("Other case")

```

```python
numbers = [4,3,1,2]
mid=len(numbers)//2
firsthalf=numbers[:mid]
secondhalf=numbers[mid:]
if sum(firsthalf)>sum(secondhalf):
    print("First half greater")
elif sum(firsthalf) == sum(secondhalf):
    print("Equal halves")
else:
    print("second half greater")
```

```python
text='hello'
if len(text)>=5:
    midindex=len(text)//2
    midchar=text[midindex:midindex+1]
    if midchar.lower() in "aeiou":
        print("Middle character is a vowel")
    else:
        print("Middle character is not a vowel")
else:
    print("string is too short")
```

```python
text = "ABZ"
if text[:1]=='A':
    if text[-1:]=='Z':
        print("valid AZ string")
    else:
        print("start with A but invalid end")
else:
    print("invalid string")
```

```python
text='12ab'
first2=text[:2]
last2=text[-2:]
if first2.isdigit():
    if last2.isalpha():
        print("Valid format")
    else:
        print("Invalid format")
else:
    print("invalid format")
```

```python
lst = [1,2,3,4,1,2]
if len(lst)%2==0:
    first2=lst[:2]
    last2=lst[-2:]
    if first2 == last2:
        print("Slice with same elements")
    else:
        print("Slice with different elements")
else:
    print("Odd length list")
```
