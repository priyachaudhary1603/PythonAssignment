```python
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender


class Student(Person):
    def __init__(self, name, age, gender):
        Person.__init__(self, name, age, gender)

    def display_student(self):
        print(self.name, self.age, self.gender)


class Academic(Student):
    def __init__(self, name, age, gender, subject, marks):
        Student.__init__(self, name, age, gender)
        self.subject = subject
        self.marks = marks

    def display_academic(self):
        print(self.name, self.age, self.gender, self.subject, self.marks)


class Sport(Student):
    def __init__(self, name, age, gender, sport_name, level):
        Student.__init__(self, name, age, gender)
        self.sport_name = sport_name
        self.level = level

    def display_sports(self):
        print(self.name, self.age, self.gender, self.sport_name, self.level)


class AllRounderStudent(Academic, Sport):
    def __init__(self, name, age, gender, subject, marks, sport_name, level):
        Academic.__init__(self, name, age, gender, subject, marks)
        Sport.__init__(self, name, age, gender, sport_name, level)

    def display(self):
        self.display_student()
        self.display_academic()
        self.display_sports()
```

```python
s1 = AllRounderStudent("Priya", 23, "F", "Math", 95, "Cricket", "State")

s1.display()
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
#Hierarchical Inheritance
class Student(Person):
    def __init__(self,name,age,rollno):
        super().__init__(name,age)
        self.rollno=rollno
#Multiple Inheritance
class Academic:
    def __init__(self,course,cgpa):
        self.course=course
        self.cgpa=cgpa
class Sports:
    def __init__(self,sport_name=None,level=None):
        self.sport_name=sport_name
        self.level=level
#Hybrid Inheritance
class AllRounderStudent(Student,Academic,Sports):
    def __init__(self,name,age,rollno,is_academic=False,
                 is_sports=False,course=None,cgpa=None,
                 sport_name=None,level=None):
        super().__init__(name,age,rollno)
        self.is_academic=is_academic
        self.is_sports=is_sports

        if self.is_academic:
            Academic.__init__(self,course,cgpa)
        if self.is_sports:
            Sports.__init__(self,sport_name,level)
            
    def display(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Roll no: {self.rollno}")

        if self.is_academic:
            print(f"Course: {self.course}")
            print(f"CGPA: {self.cgpa}")

        if self.is_sports:
            print(f"Sports name: {self.sport_name}")
            print(f"Sports level: {self.level}")

        
            
        
    
```

```python
#Academic student
s1=AllRounderStudent('Priya',22,101,is_academic=True,
                     course='IT',cgpa=7.0)
s1.display()
```

```python
from abc import ABC,abstractmethod
class BankAccount(ABC):
    def __init__(cls,customer_name,date_of_birth):
        cls.customer_name=customer_name
        cls.date_of_birth=date_of_birth
    @abstractmethod
    def deposite(self):
        pass
    def withdraw(self):
        pass
class SavingsAccount(BankAccount):
    def __init__(self,customer_name,date_of_birth,balance,amount,minimum_balance):
        super().__init__(customer_name,date_of_birth)
        self.balance=balance
        self.amount=amount
        self.minimum_balance=minimum_balance
    def deposite(self):
        self.balance=self.balance+self.amount
        print(f"The updated balance is{self.balance}")
    def withdraw(self):
        if self.amount<=(self.balance-self.minimum_balance):
            self.balance=self.balance-self.amount
            print(f"The updated balance is{self.balance}")
class CurrentAccount(BankAccount):
    def __init__(self,customer_name,date_of_birth,balance,amount,minimum_balance,overdraft):
        super().__init__(customer_name,date_of_birth)
        self.amount=amount
        self.balance=balance
        self.minimum_balance=minimum_balance
        self.overdraft=overdraft
    def deposite(self):
        self.balance=self.amount+self.balance
        print(f"The upload balance is{self.balance}")
    def withdraw(self):
        if(self.balance-self.amount>=self.overdraft):
            self.balance=self.balance-self.amount
            print(f"The updated balance is{self.balance}")
        else:
            print("Cannot update balance")


```
